```python
%pip install plotly
%pip install nbformat
import pandas as pd
import plotly.express as px
import numpy as np

# 1. Load the CSV (Make sure the path points to your new folder!)
df = pd.read_csv('/users/PCON0003/ulricclaar/desi-openspace-visualization/openspace_assets/desi_catalog_final.csv')

# 2. CRITICAL: Sample the data
df_sample = df.sample(n=5000)

# --- NEW COLOR FORMATTING BLOCK ---
# Plotly needs colors formatted as literal strings: 'rgb(R,G,B)' on a 0-255 scale.
# We multiply your 0.0-1.0 columns by 255, convert them to integers, and then to strings
r_str = (df_sample['color_r'] * 255).astype(int).astype(str)
g_str = (df_sample['color_g'] * 255).astype(int).astype(str)
b_str = (df_sample['color_b'] * 255).astype(int).astype(str)

# Stitch them together into the format Plotly expects
plotly_colors = 'rgb(' + r_str + ',' + g_str + ',' + b_str + ')'
# ----------------------------------

# 3. Create the 3D interactive plot
print("Rendering universe...")

# Notice we removed the color='object_type' argument here so Plotly doesn't override us
fig = px.scatter_3d(
    df_sample,
    x='x', y='y', z='z',
    opacity=0.9 # Upped the opacity slightly so the custom colors pop more
)

# Apply our literal custom colors and size directly to the markers
fig.update_traces(marker=dict(size=2, color=plotly_colors))

# Remove the white background to make it look more like space
fig.update_layout(template="plotly_dark")

fig.show()
```
